**Dividir la cuenta**

```python
cuenta = 72800
propina = 10
amigos = 6

#  1. Regresa la cuenta más la propina
def cuenta_propina(cuenta, propina):
  return cuenta + (cuenta * propina/100)

#  2. Regresa cuanto paga cada persona
def valor_ind(total, amigos):
  return round(total / amigos, 2)

#  3. Impresion valores
cuenta_total = cuenta_propina(cuenta, propina)
print(cuenta_total)

print(valor_ind(cuenta_total, amigos))
```

**La Caja de la Tienda**

```python
precios = [750000, 1200000, 840000, 325000, 2600000]

# Suma total
t = 0
for i in precios:
  t += i

print("La suma total de los valores es:", t)

# Si el total pasa de 5000000 aplica 10% dto
if(t > 5000000):
  d = (t*0.1)
  td = t - d

# Imprimiendo el ticket
print("Total:", t)
print("Descuento:",d)
print("A pagar", td)
```
