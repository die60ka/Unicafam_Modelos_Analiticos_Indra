**¿Te alcanza para el café?**

```python
presupuesto = 300000
precio_cafe = 4600
cafes_semana = 8

# Cuanto gasto en el mes?
gasto_mes = ((precio_cafe * cafes_semana) * 4)

# Cuanto Sobra?
sobra = presupuesto - gasto_mes

# Alcanza?
alcanza = presupuesto >= gasto_mes

# Impresiones
print(gasto_mes, sobra, alcanza)
```

**¿Tengo envío gratis?**

```python
total_c = 300
es_miembro = False
envio_gratis = False
compras = 10
nivel = ''

total = total_c * compras

if (total > 1000 or es_miembro):
  envio_gratis = True
else:
  envio_gratis = False
if (compras >= 10):
  nivel = 'Oro'
elif (compras >= 5 and compras < 10):
  nivel = 'Plata'
else:
  nivel = 'Bronce'

print(envio_gratis)
print('Nivel', nivel)
```

**Meta de Ahorro**

```python

ahorro = 900000
meta = 15000000
por_mes = 800000
meses = 0

# Bloque While
while (ahorro < meta):
  meses = meses + 1
  ahorro = ahorro + por_mes
  print('mes', meses)

# Número de meses a esperar
print()
print("Se debe esperar", meses, "meses")

```
